```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent
TABLES_DIR = PROJECT_ROOT / "reports" / "tables"

optimized_rf_metrics = pd.read_csv(
    TABLES_DIR / "optimized_random_forest_test_results.csv"
)

optimized_mlp_metrics = pd.read_csv(
    TABLES_DIR / "optimized_mlp_test_results.csv"
)
```

```python
final_model_comparison = pd.concat(
    [
        optimized_rf_metrics,
        optimized_mlp_metrics
    ],
    ignore_index=True
)

final_model_comparison
```

Prema f1 score metrici, koja je izabrana kao primarna zbog neuravnotezenosti ciljnih klasa, optimizovani MLP ostvaruje bolji rezultat.<br>
Random Forest ima visi recall i roc_auc vrednost, dok optimizovani MLP ostvaruje visu accuracy, precision i f1 vrednost.<br>
Razlike izmedju modela su male i ne ukazuju na izrazitu dominaciju jednog pristupa.
